```python
import csv

data = [
    ["transaction_id", "user_id", "product_id", "quantity", "price", "timestamp"],
    ["1001", "user_A", "prod_X", "2", "50.00", "2023-01-01T12:00:00Z"],
    ["1002", "user_B", "prod_Y", "1", "120.50", "2023-01-01T12:05:00Z"],
    ["1003", "user_C", "prod_Z", "3", "75.00", "2023-01-01T12:10:00Z"],
    ["1004", "user_D", "prod_X", "5", "50.00", "2023-01-01T12:15:00Z"],
    ["1005", "user_E", "prod_Y", "2", "120.50", "2023-01-01T12:20:00Z"]
]

with open("data.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(data)

print("CSV file created successfully!")
```

```python
import csv
import json
from pymongo import MongoClient

# Connect to MongoDB
client = MongoClient("mongodb://localhost:27017")

db = client["bigdata_db"]
collection = db["transactions"]

# Read CSV file
documents = []

with open("data.csv", "r") as f:
    reader = csv.DictReader(f)

    for row in reader:
        document = {
            "transaction_id": row["transaction_id"],
            "user_id": row["user_id"],
            "product_id": row["product_id"],
            "quantity": float(row["quantity"]),
            "price": float(row["price"]),
            "timestamp": row["timestamp"]
        }

        documents.append(document)

# Clear old test records
collection.delete_many({})

# Insert documents
result = collection.insert_many(documents)

print("=== MongoDB Output ===")
print("Database: bigdata_db")
print("Collection: transactions")
print("Total records inserted:", len(result.inserted_ids))

from bson import json_util

for document in collection.find():
    print(json_util.dumps(document))
print("Data stored successfully in MongoDB!")
```
